```python
def recommend_exercises(target, equipment, top_n=5):
    results = df[
        (df["target"] == target) &
        (df["equipment"] == equipment)
    ]

    results = results.drop_duplicates(subset=["name"])

    return results[
        ["name", "target", "equipment", "secondary_muscles", "instructions_en"]
    ].head(top_n)
```

```python
def recommend_exercises_scored(
    target,
    equipment,
    body_part=None,
    secondary_muscle=None,
    top_n=5
):
    results = df.copy()

    # Target muscle score
    results["target_score"] = (
        results["target"] == target
    ).astype(int) * 3

    # Equipment score
    results["equipment_score"] = (
        results["equipment"] == equipment
    ).astype(int) * 2

    # Body part score
    if body_part is not None:
        results["body_part_score"] = (
            results["body_part"] == body_part
        ).astype(int)
    else:
        results["body_part_score"] = 0

    # Secondary muscle score
    if secondary_muscle is not None:
        results["secondary_score"] = results["secondary_muscles"].apply(
            lambda muscles: 1 if secondary_muscle in muscles else 0
        )
    else:
        results["secondary_score"] = 0

    # Total recommendation score
    results["recommendation_score"] = (
        results["target_score"]
        + results["equipment_score"]
        + results["body_part_score"]
        + results["secondary_score"]
    )

    # Target muscle MUST match
    results = results[
        results["target"] == target
    ]

    # Remove repeated exercise names
    results = results.drop_duplicates(
        subset=["name"]
    )

    # Sort best matches first
    results = results.sort_values(
        by="recommendation_score",
        ascending=False
    )

    return results[
        [
            "name",
            "body_part",
            "target",
            "equipment",
            "secondary_muscles",
            "instructions_en",
            "instruction_steps_en",
            "recommendation_score"
        ]
    ].head(top_n)
```

```python
recommend_exercises_scored(
    target="pectorals",
    equipment="dumbbell",
    body_part="chest",
    secondary_muscle="triceps",
    top_n=10
)
```

```python
def workout_recommender():
    print("=== Workout Recommendation System ===")

    user_target = input("Enter the muscle you want to train: ").lower().strip()
    equipment_input = input("Enter available equipment: ").lower().strip()
    body_part_input = input("Enter body part: ").lower().strip()
    secondary_input = input("Enter preferred secondary muscle (optional): ").lower().strip()

    # User-friendly mappings
    target_input = target_mapping.get(user_target, user_target)
    equipment_input = equipment_mapping.get(equipment_input, equipment_input)

    if secondary_input == "":
        secondary_input = None

    # Validate inputs
    if target_input not in valid_targets:
        print("\nInvalid target muscle.")
        return

    if equipment_input not in valid_equipment:
        print("\nInvalid equipment.")
        return

    if body_part_input not in valid_body_parts:
        print("\nInvalid body part.")
        return

    # Generate recommendations
    recommendations = recommend_exercises_scored(
        target=target_input,
        equipment=equipment_input,
        body_part=body_part_input,
        secondary_muscle=secondary_input,
        top_n=5
    )

    if recommendations.empty:
        print("\nNo suitable exercises were found.")
        return

    print("\n=== Top Recommended Exercises ===\n")

    for i, row in recommendations.reset_index(drop=True).iterrows():
        print(f"{i + 1}. {row['name'].title()}")
        print(f"   Target Muscle: {row['target'].title()}")
        print(f"   Equipment: {row['equipment'].title()}")
        print(f"   Secondary Muscles: {', '.join(row['secondary_muscles']).title()}")
        print(f"   Recommendation Score: {row['recommendation_score']}")

        instructions = row.get(
            "instructions_en",
            "No instructions available."
        )

        if isinstance(instructions, list):
            instructions = " ".join(instructions)

        print(f"\n   Instructions: {instructions}")
        print("-" * 70)
```
